```python
# build_tables.py
# Generates publication-ready LaTeX tables from table-data CSV files.
#
# Run from project root:
#   python python/build_tables.py
#
# Inputs:
#   results/table_data/*.csv
#
# Outputs:
#   tables/table_01_constructs.tex
#   ...
#   tables/table_09_readiness_threshold_robustness.tex

from __future__ import annotations

from pathlib import Path
import math
import pandas as pd


# ------------------------------------------------------------
# Project paths
# ------------------------------------------------------------

try:
    # Works when running as a .py script:
    # python python/build_tables.py
    ROOT = Path(__file__).resolve().parents[1]

except NameError:
    # Works when running inside Jupyter / notebook.
    ROOT = Path.cwd()

    # If the notebook is inside the python/ folder, move one level up.
    if not (ROOT / "results" / "table_data").exists() and (ROOT.parent / "results" / "table_data").exists():
        ROOT = ROOT.parent

INPUT_DIR = ROOT / "results" / "table_data"
OUTPUT_DIR = ROOT / "tables"

if not INPUT_DIR.exists():
    raise FileNotFoundError(
        f"Could not find input directory: {INPUT_DIR}\n"
        "Run this from the project root, or set ROOT manually."
    )

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"Project root: {ROOT}")
print(f"Input directory: {INPUT_DIR}")
print(f"Output directory: {OUTPUT_DIR}")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# Basic formatting utilities
# ------------------------------------------------------------

def latex_escape_text(value: object) -> str:
    """Escape LaTeX special characters in ordinary text.

    Math strings enclosed in $...$ are preserved.
    """

    if pd.isna(value):
        return ""

    text = str(value)

    if text.startswith("$") and text.endswith("$"):
        return text

    replacements = {
        "\\": r"\textbackslash{}",
        "&": r"\&",
        "%": r"\%",
        "#": r"\#",
        "_": r"\_",
        "{": r"\{",
        "}": r"\}",
        "~": r"\textasciitilde{}",
        "^": r"\textasciicircum{}",
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    return text


def fmt_number(value: object, digits: int = 1) -> str:
    """Format numeric values with thousands separators and fixed decimals."""

    if pd.isna(value):
        return ""

    try:
        x = float(value)
    except Exception:
        return latex_escape_text(value)

    if abs(x) >= 1000:
        return f"{x:,.{digits}f}"

    return f"{x:.{digits}f}"


def fmt_integer(value: object) -> str:
    if pd.isna(value):
        return ""

    try:
        x = int(round(float(value)))
        return f"{x:,}"
    except Exception:
        return latex_escape_text(value)


def fmt_percent_from_rate(value: object, digits: int = 1) -> str:
    """Formats 1.0 as 100.0\\%."""

    if pd.isna(value):
        return ""

    try:
        x = float(value) * 100.0
        return f"{x:.{digits}f}\\%"
    except Exception:
        return latex_escape_text(value)


def fmt_percent_value(value: object, digits: int = 1) -> str:
    """Formats 46.3 as 46.3\\%."""

    if pd.isna(value):
        return ""

    try:
        x = float(value)
        return f"{x:.{digits}f}\\%"
    except Exception:
        return latex_escape_text(value)


def fmt_bool(value: object) -> str:
    if pd.isna(value):
        return ""

    try:
        return "yes" if int(float(value)) == 1 else "no"
    except Exception:
        text = str(value).strip().lower()
        if text in {"true", "1"}:
            return "yes"
        if text in {"false", "0"}:
            return "no"
        return latex_escape_text(value)


def write_tex(filename: str, content: str) -> None:
    path = OUTPUT_DIR / filename
    path.write_text(content, encoding="utf-8")
    print(f"Wrote {path.relative_to(ROOT)}")


def read_csv(name: str) -> pd.DataFrame:
    path = INPUT_DIR / name
    if not path.exists():
        raise FileNotFoundError(f"Missing required input file: {path}")
    return pd.read_csv(path)


def table_note(text: str) -> str:
    return (
        "\\begin{tablenotes}\n"
        "\\footnotesize\n"
        f"\\item Notes: {text}\n"
        "\\end{tablenotes}"
    )


def table_wrap(label: str, caption: str, body: str, note: str, size: str = "\\small") -> str:
    """Wraps tabular body in a robust table* + threeparttable environment."""

    return f"""% Auto-generated by python/build_tables.py
\\begin{{table*}}[!htbp]
\\centering
\\begin{{threeparttable}}
\\caption{{{caption}}}
\\label{{{label}}}
{size}
{body}
{table_note(note)}
\\end{{threeparttable}}
\\end{{table*}}
"""


# ------------------------------------------------------------
# Table 1
# ------------------------------------------------------------

def build_table_01() -> None:
    df = read_csv("table_01_constructs.csv").sort_values("row_order")

    rows = []
    for _, r in df.iterrows():
        rows.append(
            f"{latex_escape_text(r['construct'])} & "
            f"{latex_escape_text(r['formal_representation'])} & "
            f"{latex_escape_text(r['managerial_interpretation'])} \\\\"
        )

    body = (
        "\\begin{tabular}{p{0.23\\textwidth}p{0.23\\textwidth}p{0.47\\textwidth}}\n"
        "\\toprule\n"
        "\\textbf{Construct} & \\textbf{Formal representation} & \\textbf{Managerial interpretation} \\\\\n"
        "\\midrule\n"
        + "\n".join(rows)
        + "\n\\bottomrule\n"
        "\\end{tabular}"
    )

    note = (
        "The model does not represent full interpersonal trust or realised R\\&D performance. "
        "It represents relational confidence at the tie level and relational coordination readiness at the boundary level."
    )

    write_tex(
        "table_01_constructs.tex",
        table_wrap("tab:constructs", "Model constructs and managerial interpretation.", body, note)
    )


# ------------------------------------------------------------
# Table 2
# ------------------------------------------------------------

def build_table_02() -> None:
    df = read_csv("table_02_baseline_parameters.csv").sort_values("row_order")

    rows = []
    for _, r in df.iterrows():
        rows.append(
            f"{latex_escape_text(r['parameter'])} & "
            f"{latex_escape_text(r['meaning'])} & "
            f"{latex_escape_text(r['value'])} & "
            f"{latex_escape_text(r['managerial_rationale'])} \\\\"
        )

    body = (
        "\\begin{tabular}{p{0.13\\textwidth}p{0.25\\textwidth}p{0.12\\textwidth}p{0.42\\textwidth}}\n"
        "\\toprule\n"
        "\\textbf{Parameter} & \\textbf{Meaning} & \\textbf{Value} & \\textbf{Managerial rationale} \\\\\n"
        "\\midrule\n"
        + "\n".join(rows)
        + "\n\\bottomrule\n"
        "\\end{tabular}"
    )

    note = (
        "Parameter values define a transparent reference case rather than an empirical calibration. "
        "Robustness analyses vary the main parameters linked to translation capability, boundary-spanner load, "
        "interaction selection, and readiness thresholds."
    )

    write_tex(
        "table_02_baseline_parameters.tex",
        table_wrap("tab:baseline-parameters", "Baseline parameterization of the simulation model.", body, note, "\\scriptsize")
    )


# ------------------------------------------------------------
# Table 3
# ------------------------------------------------------------

def build_table_03() -> None:
    df = read_csv("table_03_experimental_design.csv").sort_values("row_order")

    rows = []
    for _, r in df.iterrows():
        rows.append(
            f"{latex_escape_text(r['experiment'])} & "
            f"{latex_escape_text(r['conditions'])} & "
            f"{latex_escape_text(r['managerial_question'])} \\\\"
        )

    body = (
        "\\begin{tabular}{p{0.25\\textwidth}p{0.32\\textwidth}p{0.36\\textwidth}}\n"
        "\\toprule\n"
        "\\textbf{Experiment} & \\textbf{Conditions} & \\textbf{Managerial question} \\\\\n"
        "\\midrule\n"
        + "\n".join(rows)
        + "\n\\bottomrule\n"
        "\\end{tabular}"
    )

    note = (
        "All primary comparisons are conducted at the graph level. For each graph, 50 stochastic trajectories are "
        "summarized using mean, median, $P_{90}$, and $P_{95}$ first-passage time."
    )

    write_tex(
        "table_03_experimental_design.tex",
        table_wrap("tab:experimental-design", "Experimental design and managerial questions.", body, note, "\\scriptsize")
    )


# ------------------------------------------------------------
# Table 4
# ------------------------------------------------------------

def build_table_04() -> None:
    df = read_csv("table_04_baseline_architecture.csv")

    rows = []
    for _, r in df.iterrows():
        rows.append(
            f"{latex_escape_text(r['architecture_label'])} & "
            f"{fmt_number(r['mean_T'])} & "
            f"{fmt_number(r['median_T'])} & "
            f"{fmt_number(r['p90_T'])} & "
            f"{fmt_number(r['p95_T'])} & "
            f"{fmt_percent_from_rate(r['convergence_rate'], 0)} \\\\"
        )

    body = (
        "\\begin{tabular}{p{0.30\\textwidth}rrrrr}\n"
        "\\toprule\n"
        "\\textbf{Architecture} & \\textbf{Mean $T$} & \\textbf{Median $T$} & "
        "\\textbf{$P_{90}$} & \\textbf{$P_{95}$} & \\textbf{Convergence} \\\\\n"
        "\\midrule\n"
        + "\n".join(rows)
        + "\n\\bottomrule\n"
        "\\end{tabular}"
    )

    note = (
        "Values are graph-level summaries of first-passage time to relational coordination readiness. "
        "Each architecture uses $NG=50$ graph realizations and $NT=50$ trajectories per graph. "
        "Random bridging does not clearly outperform the modular baseline, whereas boundary spanning substantially "
        "reduces readiness time and upper-tail delay."
    )

    write_tex(
        "table_04_baseline_architecture.tex",
        table_wrap("tab:baseline-architecture-results", "Baseline architecture comparison.", body, note)
    )


# ------------------------------------------------------------
# Table 5
# ------------------------------------------------------------

def build_table_05() -> None:
    df = read_csv("table_05_mechanism_decomposition.csv")

    comparison_order = [
        "RB_low_minus_BS_low",
        "BS_low_minus_BS_high",
        "RB_low_minus_BS_high",
    ]

    rows = []

    for c_i, comp in enumerate(comparison_order):
        sub = df[df["comparison_code"] == comp]

        if c_i > 0:
            rows.append("\\addlinespace")

        for _, r in sub.iterrows():
            reduction = ""
            if not pd.isna(r["reduction_pct"]):
                reduction = fmt_percent_value(r["reduction_pct"])

            interval = f"[{fmt_number(r['ci_low'])}, {fmt_number(r['ci_high'])}]"

            comparison_label = latex_escape_text(r["comparison_label"])

            rows.append(
                f"{comparison_label} & "
                f"{latex_escape_text(r['statistic'])} & "
                f"{latex_escape_text(r['condition_x'])} & "
                f"{latex_escape_text(r['condition_y'])} & "
                f"{fmt_number(r['x_value'])} & "
                f"{fmt_number(r['y_value'])} & "
                f"{fmt_number(r['difference_x_minus_y'])} & "
                f"{interval} & "
                f"{reduction} \\\\"
            )

    body = (
        "\\begin{tabular}{p{0.20\\textwidth}p{0.10\\textwidth}p{0.08\\textwidth}p{0.08\\textwidth}rrrrr}\n"
        "\\toprule\n"
        "\\textbf{Comparison} & \\textbf{Statistic} & \\textbf{X} & \\textbf{Y} & "
        "\\textbf{X value} & \\textbf{Y value} & \\textbf{X--Y} & "
        "\\textbf{Bootstrap interval} & \\textbf{Reduction} \\\\\n"
        "\\midrule\n"
        + "\n".join(rows)
        + "\n\\bottomrule\n"
        "\\end{tabular}"
    )

    note = (
        "$RB_{low}$ is random bridging with ordinary cross-boundary success probability. "
        "$BS_{low}$ is boundary spanning without translation advantage, with $\\pi_{BS}=\\pi_{out}=0.55$. "
        "$BS_{high}$ is boundary spanning with translation advantage, with $\\pi_{BS}=0.65$. "
        "Positive differences indicate that condition Y has lower first-passage time than condition X. "
        "Negative differences indicate that condition Y is slower than condition X. Intervals are graph-level bootstrap intervals."
    )

    write_tex(
        "table_05_mechanism_decomposition.tex",
        table_wrap(
            "tab:mechanism-decomposition",
            "Mechanism decomposition: concentration, translation capability, and total boundary-spanning effect.",
            body,
            note,
            "\\scriptsize",
        )
    )


# ------------------------------------------------------------
# Table 6
# ------------------------------------------------------------

def build_table_06() -> None:
    df = read_csv("table_06_translation_capability.csv").sort_values("pi_BS")

    rows = []
    for _, r in df.iterrows():
        rows.append(
            f"{fmt_number(r['pi_BS'], 2)} & "
            f"{fmt_number(r['mean_T'])} & "
            f"{fmt_number(r['median_T'])} & "
            f"{fmt_number(r['p90_T'])} & "
            f"{fmt_number(r['p95_T'])} \\\\"
        )

    body = (
        "\\begin{tabular}{rrrrr}\n"
        "\\toprule\n"
        "\\textbf{$\\pi_{BS}$} & \\textbf{Mean $T$} & \\textbf{Median $T$} & "
        "\\textbf{$P_{90}$} & \\textbf{$P_{95}$} \\\\\n"
        "\\midrule\n"
        + "\n".join(rows)
        + "\n\\bottomrule\n"
        "\\end{tabular}"
    )

    note = (
        "The table reports graph-level first-passage times under the boundary-spanning architecture while varying "
        "the boundary-spanning success probability $\\pi_{BS}$ and holding $\\pi_{out}=0.55$. "
        "The monotonic decline supports translation capability as the switching mechanism."
    )

    write_tex(
        "table_06_translation_capability.tex",
        table_wrap("tab:translation-capability-grid", "Translation-capability robustness.", body, note)
    )


# ------------------------------------------------------------
# Table 7
# ------------------------------------------------------------

def build_table_07() -> None:
    df = read_csv("table_07_boundary_spanner_load.csv").sort_values("b")

    rows = []
    for _, r in df.iterrows():
        rows.append(
            f"{fmt_integer(r['b'])} & "
            f"{fmt_number(r['load_per_spanner'], 1)} & "
            f"{fmt_number(r['mean_T'])} & "
            f"{fmt_number(r['median_T'])} & "
            f"{fmt_number(r['p90_T'])} & "
            f"{fmt_number(r['p95_T'])} \\\\"
        )

    body = (
        "\\begin{tabular}{rrrrrr}\n"
        "\\toprule\n"
        "\\textbf{$b$} & \\textbf{Load $k/(2b)$} & \\textbf{Mean $T$} & "
        "\\textbf{Median $T$} & \\textbf{$P_{90}$} & \\textbf{$P_{95}$} \\\\\n"
        "\\midrule\n"
        + "\n".join(rows)
        + "\n\\bottomrule\n"
        "\\end{tabular}"
    )

    note = (
        "The cross-boundary tie budget is fixed at $k=12$. Load is measured as $k/(2b)$, where $b$ is the number "
        "of boundary spanners per side. Within the tested range, lower load monotonically reduces mean, median, "
        "$P_{90}$, and $P_{95}$ readiness time."
    )

    write_tex(
        "table_07_boundary_spanner_load.tex",
        table_wrap("tab:boundary-spanner-load", "Boundary-spanner load and readiness delay.", body, note)
    )


# ------------------------------------------------------------
# Table 8
# ------------------------------------------------------------

def build_table_08() -> None:
    df = read_csv("table_08_selection_rule_robustness.csv")

    condition_order = {"RB_low": 1, "BS_low": 2, "BS_high": 3}
    rule_order = {"agent_first": 1, "edge_uniform": 2}

    df["condition_order"] = df["condition"].map(condition_order)
    df["rule_order"] = df["selection_rule"].map(rule_order)
    df = df.sort_values(["condition_order", "rule_order"])

    rows = []
    for _, r in df.iterrows():
        rule = "Agent-first" if r["selection_rule"] == "agent_first" else "Edge-uniform"

        rows.append(
            f"{latex_escape_text(r['condition'])} & "
            f"{rule} & "
            f"{fmt_number(r['mean_T'])} & "
            f"{fmt_number(r['median_T'])} & "
            f"{fmt_number(r['p90_T'])} & "
            f"{fmt_number(r['p95_T'])} \\\\"
        )

    body = (
        "\\begin{tabular}{p{0.18\\textwidth}p{0.22\\textwidth}rrrr}\n"
        "\\toprule\n"
        "\\textbf{Condition} & \\textbf{Selection rule} & \\textbf{Mean $T$} & "
        "\\textbf{Median $T$} & \\textbf{$P_{90}$} & \\textbf{$P_{95}$} \\\\\n"
        "\\midrule\n"
        + "\n".join(rows)
        + "\n\\bottomrule\n"
        "\\end{tabular}"
    )

    note = (
        "Agent-first selection represents scarce actor-level interaction capacity. Edge-uniform selection activates ties uniformly "
        "and therefore weakens actor-level overload. Under edge-uniform selection, the concentration penalty in $BS_{low}$ "
        "largely disappears, while translation capability remains strongly beneficial."
    )

    write_tex(
        "table_08_selection_rule_robustness.tex",
        table_wrap("tab:appendix-selection-rule-robustness", "Interaction-selection robustness: agent-first versus edge-uniform activation.", body, note)
    )


# ------------------------------------------------------------
# Table 9
# ------------------------------------------------------------

def build_table_09() -> None:
    df = read_csv("table_09_readiness_threshold_robustness.csv")

    scenario_order = {
        "TH_075_Q_080": 1,
        "TH_080_Q_070": 2,
        "TH_080_Q_080": 3,
        "TH_080_Q_090": 4,
        "TH_085_Q_080": 5,
    }

    df["scenario_order"] = df["scenario_code"].map(scenario_order)
    df = df.sort_values("scenario_order")

    rows = []
    for _, r in df.iterrows():
        rows.append(
            f"{fmt_number(r['theta'], 2)} & "
            f"{fmt_number(r['q'], 2)} & "
            f"{fmt_percent_from_rate(r['BS_low_convergence_rate'])} & "
            f"{fmt_percent_from_rate(r['BS_high_convergence_rate'])} & "
            f"{fmt_number(r['BS_low_p95_T'])} & "
            f"{fmt_number(r['BS_high_p95_T'])} & "
            f"{fmt_percent_value(r['p95_reduction_pct'])} \\\\"
        )

    body = (
        "\\begin{tabular}{rrrrrrr}\n"
        "\\toprule\n"
        "\\textbf{$\\theta$} & \\textbf{$q$} & \\textbf{$BS_{low}$ conv.} & "
        "\\textbf{$BS_{high}$ conv.} & \\textbf{$BS_{low}$ $P_{95}$} & "
        "\\textbf{$BS_{high}$ $P_{95}$} & \\textbf{Reduction} \\\\\n"
        "\\midrule\n"
        + "\n".join(rows)
        + "\n\\bottomrule\n"
        "\\end{tabular}"
    )

    note = (
        "$\\theta$ is the tie-level readiness threshold and $q$ is the boundary-level readiness threshold. "
        "The baseline condition is $\\theta=0.80$ and $q=0.80$. The last row includes censoring at "
        "$T_{\\max}=50{,}000$ because $BS_{low}$ frequently fails to reach readiness under the hardest tie-level threshold. "
        "Across all scenarios, translation-capable boundary spanning remains faster and more reliable than concentration without translation capability."
    )

    write_tex(
        "table_09_readiness_threshold_robustness.tex",
        table_wrap("tab:appendix-threshold-robustness", "Readiness-threshold robustness.", body, note, "\\scriptsize")
    )


# ------------------------------------------------------------
# Main
# ------------------------------------------------------------

def main() -> None:
    print("Building LaTeX tables from CSV table data...")
    print(f"Input directory: {INPUT_DIR.relative_to(ROOT)}")
    print(f"Output directory: {OUTPUT_DIR.relative_to(ROOT)}")

    build_table_01()
    build_table_02()
    build_table_03()
    build_table_04()
    build_table_05()
    build_table_06()
    build_table_07()
    build_table_08()
    build_table_09()

    print("\nAll LaTeX tables generated successfully.")


if __name__ == "__main__":
    main()
```
