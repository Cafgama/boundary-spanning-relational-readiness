```python
# build_figures.py
# Generates clean publication-ready figures from the figure-data CSV files.
#
# Run from the project root:
#   python python/build_figures.py
#
# Inputs:
#   results/figure_data/*.csv
#
# Outputs:
#   figures/*.pdf
#   figures/*.png

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Project paths
# ------------------------------------------------------------

try:
    ROOT = Path(__file__).resolve().parents[1]
except NameError:
    ROOT = Path.cwd()
    if not (ROOT / "results" / "figure_data").exists() and (ROOT.parent / "results" / "figure_data").exists():
        ROOT = ROOT.parent

INPUT_DIR = ROOT / "results" / "figure_data"
OUTPUT_DIR = ROOT / "figures"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

if not INPUT_DIR.exists():
    raise FileNotFoundError(f"Could not find figure-data directory: {INPUT_DIR}")

# ------------------------------------------------------------
# Style
# ------------------------------------------------------------

def get_batlow_colors(n):
    """
    Returns n colors sampled from Batlow if cmcrameri is available.
    Falls back to a Batlow-like palette otherwise.
    """
    try:
        import cmcrameri.cm as cmc
        cmap = cmc.batlow
        vals = np.linspace(0.12, 0.88, n)
        return [cmap(v) for v in vals]
    except Exception:
        fallback = [
            "#243B6B",
            "#2B5C8A",
            "#2F7E8E",
            "#49A07A",
            "#84BE5A",
            "#C9D54A",
            "#F2B13E",
            "#E47A2E",
        ]
        if n <= len(fallback):
            return fallback[:n]
        vals = np.linspace(0.1, 0.9, n)
        return [plt.cm.viridis(v) for v in vals]


def apply_style():
    plt.rcParams.update({
        "font.size": 11,
        "axes.titlesize": 14,
        "axes.labelsize": 12,
        "legend.fontsize": 10,
        "xtick.labelsize": 10,
        "ytick.labelsize": 10,
        "figure.dpi": 140,
        "savefig.dpi": 300,
        "axes.spines.top": False,
        "axes.spines.right": False,
        "axes.grid": True,
        "grid.alpha": 0.25,
        "grid.linestyle": "--",
        "grid.linewidth": 0.6,
    })


def save_figure(fig, name):
    pdf_path = OUTPUT_DIR / f"{name}.pdf"
    png_path = OUTPUT_DIR / f"{name}.png"
    fig.tight_layout()
    fig.savefig(pdf_path, bbox_inches="tight")
    fig.savefig(png_path, bbox_inches="tight")
    plt.close(fig)
    print(f"Wrote {pdf_path.relative_to(ROOT)}")
    print(f"Wrote {png_path.relative_to(ROOT)}")


def fmt_axis(ax):
    ax.spines["left"].set_alpha(0.6)
    ax.spines["bottom"].set_alpha(0.6)


# ------------------------------------------------------------
# Figure 02 — architectures comparison
# ------------------------------------------------------------


def build_fig_02():
    import numpy as np
    import matplotlib.pyplot as plt

    batlow = get_batlow_colors(7)

    color_u = batlow[1]
    color_i = batlow[5]
    color_cross = "0.35"
    color_spanner = batlow[6]

    # ------------------------------------------------------------
    # Fixed node positions
    # ------------------------------------------------------------

    n_side = 10

    y_positions = np.linspace(0.1, 0.9, n_side)

    U = [(0.25, y) for y in y_positions]
    I = [(0.75, y) for y in y_positions]

    # boundary spanners: chosen fixed nodes
    bs_u = [2, 7]
    bs_i = [2, 7]

    # ------------------------------------------------------------
    # Internal ties: simple nearest-neighbor plus a few shortcuts
    # ------------------------------------------------------------

    internal_edges_u = [(i, i + 1) for i in range(n_side - 1)] + [(1, 4), (4, 7), (2, 6)]
    internal_edges_i = [(i, i + 1) for i in range(n_side - 1)] + [(1, 4), (4, 7), (2, 6)]

    # ------------------------------------------------------------
    # Cross-boundary ties for each panel
    # ------------------------------------------------------------

   # Baseline: sparse unmanaged cross-boundary contact.
    baseline_cross = [
        (1, 2),
        (5, 4),
        (8, 7),
    ]
    
    # Random bridging: same cross-boundary budget used in the controlled
    # comparison, but distributed without deliberate boundary-spanner roles.
    random_cross = [
        (0, 1),
        (1, 7),
        (2, 4),
        (3, 9),
        (4, 0),
        (5, 8),
        (6, 2),
        (7, 5),
        (8, 3),
        (9, 6),
        (0, 8),
        (9, 1),
    ]
    
    # Boundary spanning: same cross-boundary budget as random bridging,
    # but each cross-boundary tie is incident to a designated boundary spanner.
    # With b = 2 per side and k = 12, each of the four boundary spanners
    # carries three cross-boundary ties in this schematic.
    bs_cross = [
        # University-side boundary spanners carry ties to industry actors.
        (2, 0),
        (2, 3),
        (2, 5),
        (7, 4),
        (7, 6),
        (7, 9),
    
        # Industry-side boundary spanners carry ties to university actors.
        (0, 2),
        (3, 2),
        (5, 2),
        (4, 7),
        (6, 7),
        (9, 7),
]

    # ------------------------------------------------------------
    # Helper to draw one panel
    # ------------------------------------------------------------

    def draw_panel(ax, title, cross_edges, highlight_spanners=False):
        # internal ties U
        for a, b in internal_edges_u:
            xa, ya = U[a]
            xb, yb = U[b]
            ax.plot([xa, xb], [ya, yb], linewidth=1.1, color=color_u, alpha=0.55)

        # internal ties I
        for a, b in internal_edges_i:
            xa, ya = I[a]
            xb, yb = I[b]
            ax.plot([xa, xb], [ya, yb], linewidth=1.1, color=color_i, alpha=0.55)

        # cross-boundary ties
        for a, b in cross_edges:
            xa, ya = U[a]
            xb, yb = I[b]
            ax.plot([xa, xb], [ya, yb], linewidth=1.3, color=color_cross, alpha=0.9)

        # nodes U
        for idx, (x, y) in enumerate(U):
            if highlight_spanners and idx in bs_u:
                ax.scatter(x, y, s=90, color=color_spanner, edgecolor="black", linewidth=0.6, zorder=3)
            else:
                ax.scatter(x, y, s=55, color=color_u, edgecolor="white", linewidth=0.5, zorder=3)

        # nodes I
        for idx, (x, y) in enumerate(I):
            if highlight_spanners and idx in bs_i:
                ax.scatter(x, y, s=90, color=color_spanner, edgecolor="black", linewidth=0.6, zorder=3)
            else:
                ax.scatter(x, y, s=55, color=color_i, edgecolor="white", linewidth=0.5, zorder=3)

        ax.text(0.25, 0.97, "University", ha="center", va="top", fontsize=11)
        ax.text(0.75, 0.97, "Industry", ha="center", va="top", fontsize=11)

        ax.set_title(title)
        ax.set_xlim(0.05, 0.95)
        ax.set_ylim(0.02, 1.0)
        ax.set_xticks([])
        ax.set_yticks([])
        for spine in ax.spines.values():
            spine.set_visible(False)

    # ------------------------------------------------------------
    # Build figure
    # ------------------------------------------------------------

    fig, axes = plt.subplots(1, 3, figsize=(13.5, 4.6))

    draw_panel(axes[0], "Baseline modular network", baseline_cross, highlight_spanners=False)
    draw_panel(axes[1], "Random bridging", random_cross, highlight_spanners=False)
    draw_panel(axes[2], "Boundary spanning", bs_cross, highlight_spanners=True)

    fig.suptitle("Network architectures used in the computational experiments", y=1.02, fontsize=14)

    save_figure(fig, "fig_02_network_architectures")





# ------------------------------------------------------------
# Figure 03 — Baseline architecture comparison
# ------------------------------------------------------------

def build_fig_03():
    df = pd.read_csv(INPUT_DIR / "fig_03_baseline_architecture.csv")

    order = [
        "Baseline modular network",
        "Random bridging",
        "Boundary spanning",
    ]
    df["architecture_label"] = pd.Categorical(df["architecture_label"], categories=order, ordered=True)
    df = df.sort_values("architecture_label")

    metrics = ["mean_T", "median_T", "p90_T", "p95_T"]
    metric_labels = ["Mean", "Median", "P90", "P95"]
    colors = get_batlow_colors(len(metrics))

    y = np.arange(len(df))
    h = 0.18
    offsets = np.array([-1.5, -0.5, 0.5, 1.5]) * h

    fig, ax = plt.subplots(figsize=(10, 5.4))

    for i, (m, label, color) in enumerate(zip(metrics, metric_labels, colors)):
        ax.barh(y + offsets[i], df[m], height=h, label=label, color=color)

    ax.set_yticks(y)
    ax.set_yticklabels(df["architecture_label"])
    ax.invert_yaxis()
    ax.set_xlabel("Time to relational readiness")
    ax.set_title("Baseline architecture comparison")
    ax.legend(frameon=False, ncol=4, loc="lower right")
    fmt_axis(ax)

    save_figure(fig, "fig_03_baseline_architecture")


# ------------------------------------------------------------
# Figure 04 — Mechanism decomposition
# ------------------------------------------------------------

def build_fig_04():
    df = pd.read_csv(INPUT_DIR / "fig_04_mechanism_decomposition.csv")

    comp_order = [
        "RB_low_minus_BS_low",
        "BS_low_minus_BS_high",
        "RB_low_minus_BS_high",
    ]

    stat_order = ["Mean T", "Median T", "P90", "P95"]

    comp_short = {
        "RB_low_minus_BS_low": "Concentration without translation",
        "BS_low_minus_BS_high": "Translation capability",
        "RB_low_minus_BS_high": "Total boundary-spanning effect",
    }

    df["comparison_code"] = pd.Categorical(
        df["comparison_code"],
        categories=comp_order,
        ordered=True,
    )

    df["statistic"] = pd.Categorical(
        df["statistic"],
        categories=stat_order,
        ordered=True,
    )

    df = df.sort_values(["comparison_code", "statistic"]).reset_index(drop=True)

    df["label"] = df.apply(
        lambda r: f"{comp_short[r['comparison_code']]} | {r['statistic']}",
        axis=1,
    )

    batlow = get_batlow_colors(7)

    comp_colors = {
        "RB_low_minus_BS_low": batlow[1],
        "BS_low_minus_BS_high": batlow[4],
        "RB_low_minus_BS_high": batlow[6],
    }

    y = np.arange(len(df))

    fig, ax = plt.subplots(figsize=(11.5, 7.4))

    for i, row in df.iterrows():
        diff = row["difference_x_minus_y"]
        lo = row["ci_low"]
        hi = row["ci_high"]
        color = comp_colors[row["comparison_code"]]

        ax.errorbar(
            diff,
            i,
            xerr=[[diff - lo], [hi - diff]],
            fmt="o",
            color=color,
            ecolor=color,
            elinewidth=1.8,
            capsize=3,
            markersize=6,
            zorder=3,
        )

    # Light zero-reference line. This is not a second axis.
    ax.axvline(
        0,
        color="0.45",
        linewidth=0.9,
        linestyle="--",
        alpha=0.75,
        zorder=1,
    )

    ax.set_yticks(y)
    ax.set_yticklabels(df["label"])
    ax.invert_yaxis()

    ax.set_xlabel("Readiness-time difference, X minus Y; positive values mean Y is faster")
    ax.set_title("Mechanism decomposition of boundary-spanning effects")

    # Color-code legend.
    from matplotlib.lines import Line2D

    legend_handles = [
        Line2D(
            [0],
            [0],
            marker="o",
            linestyle="None",
            markersize=7,
            color=comp_colors["RB_low_minus_BS_low"],
            label="Concentration without translation",
        ),
        Line2D(
            [0],
            [0],
            marker="o",
            linestyle="None",
            markersize=7,
            color=comp_colors["BS_low_minus_BS_high"],
            label="Translation capability",
        ),
        Line2D(
            [0],
            [0],
            marker="o",
            linestyle="None",
            markersize=7,
            color=comp_colors["RB_low_minus_BS_high"],
            label="Total boundary-spanning effect",
        ),
        Line2D(
            [0],
            [0],
            linestyle="--",
            linewidth=1.0,
            color="0.45",
            label="No difference",
        ),
    ]

    ax.legend(
        handles=legend_handles,
        frameon=False,
        loc="upper right",
        title="Comparison",
        title_fontsize=10,
    )

    fmt_axis(ax)

    save_figure(fig, "fig_04_mechanism_decomposition")

# ------------------------------------------------------------
# Figure 05 — Tail-risk profile
# ------------------------------------------------------------

def build_fig_05():
    df = pd.read_csv(INPUT_DIR / "fig_05_tail_risk_profile.csv")

    order = ["RB_low", "BS_low", "BS_high"]
    labels = {
        "RB_low": "Random bridging, ordinary translation",
        "BS_low": "Boundary spanning, no translation advantage",
        "BS_high": "Boundary spanning, translation advantage",
    }

    df["condition"] = pd.Categorical(df["condition"], categories=order, ordered=True)
    df = df.sort_values("condition")

    x = np.arange(4)
    xlabels = ["Mean", "Median", "P90", "P95"]
    metrics = ["mean_T", "median_T", "p90_T", "p95_T"]

    colors = get_batlow_colors(3)

    fig, ax = plt.subplots(figsize=(8.5, 5.6))

    for color, (_, row) in zip(colors, df.iterrows()):
        y = [row[m] for m in metrics]
        ax.plot(x, y, marker="o", linewidth=2.2, markersize=6, color=color, label=labels[row["condition"]])

    ax.set_xticks(x)
    ax.set_xticklabels(xlabels)
    ax.set_ylabel("Time to relational readiness")
    ax.set_title("Tail-risk profile across mechanism conditions")
    ax.legend(frameon=False, loc="upper left")
    fmt_axis(ax)

    save_figure(fig, "fig_05_tail_risk_profile")


# ------------------------------------------------------------
# Figure 06 — Translation capability curve
# ------------------------------------------------------------

def build_fig_06():
    df = pd.read_csv(INPUT_DIR / "fig_06_translation_capability_curve.csv").sort_values("pi_BS")

    x = df["pi_BS"].values
    metrics = ["mean_T", "median_T", "p90_T", "p95_T"]
    labels = ["Mean", "Median", "P90", "P95"]
    colors = get_batlow_colors(4)

    fig, ax = plt.subplots(figsize=(8.5, 5.6))

    for m, label, color in zip(metrics, labels, colors):
        ax.plot(x, df[m], marker="o", linewidth=2.0, markersize=6, color=color, label=label)

    ax.set_xlabel(r"Boundary-spanning success probability $\pi_{BS}$")
    ax.set_ylabel("Time to relational readiness")
    ax.set_title("Translation capability reduces readiness delay")
    ax.legend(frameon=False)
    fmt_axis(ax)

    save_figure(fig, "fig_06_translation_capability")


# ------------------------------------------------------------
# Figure 07 — Boundary-spanner load
# ------------------------------------------------------------

def build_fig_07():
    df = pd.read_csv(INPUT_DIR / "fig_07_boundary_spanner_load.csv").sort_values("b")

    x = df["b"].values
    metrics = ["mean_T", "median_T", "p90_T", "p95_T"]
    labels = ["Mean", "Median", "P90", "P95"]
    colors = get_batlow_colors(4)

    fig, ax = plt.subplots(figsize=(8.5, 5.6))

    for m, label, color in zip(metrics, labels, colors):
        ax.plot(x, df[m], marker="o", linewidth=2.0, markersize=6, color=color, label=label)

    ax.set_xlabel("Boundary spanners per side (b)")
    ax.set_ylabel("Time to relational readiness")
    ax.set_title("Greater boundary-spanning capacity reduces delay")
    ax.legend(frameon=False)
    fmt_axis(ax)

    save_figure(fig, "fig_07_boundary_spanner_load")


# ------------------------------------------------------------
# Figure 08 — Selection-rule robustness
# ------------------------------------------------------------

def build_fig_08():
    df = pd.read_csv(INPUT_DIR / "fig_08_selection_rule_robustness.csv")

    condition_order = ["RB_low", "BS_low", "BS_high"]
    condition_labels = {
        "RB_low": "Random bridging\nordinary translation",
        "BS_low": "Boundary spanning\nno translation advantage",
        "BS_high": "Boundary spanning\ntranslation advantage",
    }

    color_map = {
        "RB_low": get_batlow_colors(7)[1],
        "BS_low": get_batlow_colors(7)[3],
        "BS_high": get_batlow_colors(7)[6],
    }

    fig, ax = plt.subplots(figsize=(8.8, 5.8))

    x = np.array([0, 1])
    xlabels = ["Agent-first", "Edge-uniform"]

    for cond in condition_order:
        sub = df[df["condition"] == cond].copy()
        sub["selection_rule"] = pd.Categorical(sub["selection_rule"], categories=["agent_first", "edge_uniform"], ordered=True)
        sub = sub.sort_values("selection_rule")

        y = sub["p95_T"].values
        ax.plot(
            x,
            y,
            marker="o",
            linewidth=2.3,
            markersize=7,
            color=color_map[cond],
            label=condition_labels[cond],
        )

    ax.set_xticks(x)
    ax.set_xticklabels(xlabels)
    ax.set_ylabel("P95 time to readiness")
    ax.set_title("Selection-rule robustness (tail-risk focus)")
    ax.legend(frameon=False, loc="upper right")
    fmt_axis(ax)

    save_figure(fig, "fig_08_selection_rule_robustness")


# ------------------------------------------------------------
# Figure 09 — Readiness-threshold robustness
# ------------------------------------------------------------

def build_fig_09():
    df = pd.read_csv(INPUT_DIR / "fig_09_readiness_threshold_robustness.csv")

    order = [
        "TH_075_Q_080",
        "TH_080_Q_070",
        "TH_080_Q_080",
        "TH_080_Q_090",
        "TH_085_Q_080",
    ]

    labels = {
        "TH_075_Q_080": "(0.75, 0.80)",
        "TH_080_Q_070": "(0.80, 0.70)",
        "TH_080_Q_080": "(0.80, 0.80)",
        "TH_080_Q_090": "(0.80, 0.90)",
        "TH_085_Q_080": "(0.85, 0.80)",
    }

    df["scenario_code"] = pd.Categorical(df["scenario_code"], categories=order, ordered=True)
    df = df.sort_values("scenario_code")

    x = np.arange(len(df))
    width = 0.36
    colors = get_batlow_colors(2)

    fig, ax = plt.subplots(figsize=(10.2, 5.8))

    ax.bar(
        x - width / 2,
        df["BS_low_p95_T"],
        width=width,
        color=colors[0],
        label="BS low",
    )

    ax.bar(
        x + width / 2,
        df["BS_high_p95_T"],
        width=width,
        color=colors[1],
        label="BS high",
    )

    ax.set_xticks(x)
    ax.set_xticklabels([labels[s] for s in df["scenario_code"]])

    ax.set_xlabel("Readiness threshold scenario (theta, q)")
    ax.set_ylabel("P95 time to readiness")
    ax.set_title("Readiness-threshold robustness")

    ax.legend(frameon=False)

    fmt_axis(ax)

    save_figure(fig, "fig_09_readiness_threshold_robustness")

# ------------------------------------------------------------
# Main
# ------------------------------------------------------------

def main():
    apply_style()

    print("Building publication-ready figures...")
    print(f"Input directory: {INPUT_DIR}")
    print(f"Output directory: {OUTPUT_DIR}")

    build_fig_02()
    build_fig_03()
    build_fig_04()
    build_fig_05()
    build_fig_06()
    build_fig_07()
    build_fig_08()
    build_fig_09()

    print("\nAll figures generated successfully.")


if __name__ == "__main__":
    main()
```
